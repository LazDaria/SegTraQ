## Creating spatialdata objects using method-specific readers

Code from these readers has to be integrated into `create_spatialdata` later.

#### 10x Xenium

```python
%load_ext autoreload
%autoreload 2
```

```python
from pathlib import Path

import segtraq as sq

base_path = Path("/home/lazic/src/SegTraQ/tests/data")
xenium_sdata = sq.fs.read_xenium(base_path / "xenium")
xenium_sdata
```

```python
sq.fs.validate_spatialdata(xenium_sdata, data_key="scale0/image")
```

```python
import spatialdata_plot
import matplotlib.pyplot as plt
import dask

axes = plt.subplots(1, 2, figsize=(10, 5), constrained_layout=True)[1].flatten()

#Dapi image
xenium_sdata.pl.render_images("morphology_focus").pl.show(
    ax=axes[0], title="DAPI image", coordinate_systems="global"
)

# Plot - Cell boundaries
xenium_sdata.tables["table"].obs["region"] = "cell_labels"
xenium_sdata.set_table_annotates_spatialelement("table", region="cell_labels")


with dask.config.set(scheduler="single-threaded"):
    xenium_sdata.pl.render_shapes("cell_labels", color="label_id").pl.show(
        ax=axes[1], title="10x Xenium: Cell mask - boundaries", coordinate_systems="global"
    )
```

```python
xenium_sdata.write(base_path / "xenium_sdata.zarr", overwrite=True)
```
